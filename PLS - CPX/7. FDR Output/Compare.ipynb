```python
import pandas as pd

# Read files
ananya_df = pd.read_csv("final_edges_with_fdr.csv", header=None, dtype=str, low_memory=False)
lauren_df = pd.read_csv("lauren_fdr_table.csv", header=None, dtype=str, low_memory=False)

# Strip whitespace everywhere
ananya_df = ananya_df.apply(lambda s: s.str.strip())
lauren_df = lauren_df.apply(lambda s: s.str.strip())

# Remove -P suffix from gene column in Ananya (skip header row if present)
ananya_df.loc[1:, 0] = ananya_df.loc[1:, 0].str.replace(r"-P$", "", regex=True)

# Rename first two columns for clarity
ananya_df = ananya_df.rename(columns={0: "gene", 1: "metabolite"})
lauren_df = lauren_df.rename(columns={0: "gene", 1: "metabolite"})
print(len(ananya_df))
print(len(lauren_df))


# Create tuple keys (gene, metabolite)
ananya_keys = set(zip(ananya_df["gene"], ananya_df["metabolite"]))
lauren_keys = set(zip(lauren_df["gene"], lauren_df["metabolite"]))

# Find mismatches
missing_in_lauren = ananya_keys - lauren_keys
missing_in_ananya = lauren_keys - ananya_keys

# Convert mismatches to DataFrames for printing
missing_in_lauren_df = pd.DataFrame(
    list(missing_in_lauren), columns=["gene", "metabolite"]
)

missing_in_ananya_df = pd.DataFrame(
    list(missing_in_ananya), columns=["gene", "metabolite"]
)

# Rearrange Lauren to match Ananya ordering
lauren_reordered = lauren_df.merge(
    ananya_df[["gene", "metabolite"]],
    on=["gene", "metabolite"],
    how="inner"
)

# Output
print("=== Pairs in Ananya but NOT in Lauren ===")
print(missing_in_lauren_df)

print("\n=== Pairs in Lauren but NOT in Ananya ===")
print(missing_in_ananya_df)
```
