```python
import pandas as pd
import re
import os

def combine(metabolomics, transcriptomics, output_file):

    # Define folder and file path
    folder_path = "4. Combine Data Tables"
    os.makedirs(folder_path, exist_ok=True)
    output_file_path = os.path.join(folder_path, output_file)

    # Read metabolomics with no header
    meta = pd.read_csv(metabolomics, header=None)
    tran = pd.read_csv(transcriptomics, header=None)


    #The first column is the same
    meta.columns = meta.iloc[0]
    meta = meta[1:].reset_index(drop=True)
    
    tran.columns = tran.iloc[0]
    tran = tran[1:].reset_index(drop=True)
    
    combined = pd.concat([meta, tran], axis=0, join="outer", ignore_index=True)

    combined.to_csv(output_file_path, index=False, header=True)


# Example call
combine("2. Filter by Choroid Plexus/DSS_CPX_filter.csv", "3. Rename Metabolomics/metabolomics_pls_dss.csv", "dss_combined.csv")
combine("2. Filter by Choroid Plexus/VECPAC_CPX_filter.csv", "3. Rename Metabolomics/metabolomics_pls_vecpac.csv", "vecpac_combined.csv")
combine("2. Filter by Choroid Plexus/LPS_CPX_filter.csv", "3. Rename Metabolomics/metabolomics_pls_lps.csv", "lps_combined.csv")
```

Create Combined Table

```python
import pandas as pd

dss = pd.read_csv("./4. Combine Data Tables/dss_combined.csv", header=None)
lps = pd.read_csv("./4. Combine Data Tables/lps_combined.csv", header=None)
vecpac = pd.read_csv("./4. Combine Data Tables/vecpac_combined.csv", header=None)

# Drop the first column of lps and vecpac
lps_trimmed = lps.iloc[:, 1:]
vecpac_trimmed = vecpac.iloc[:, 1:]

# Concatenate side-by-side with dss
combined = pd.concat([dss, lps_trimmed, vecpac_trimmed], axis=1)

combined.to_csv("./4. Combine Data Tables/all_combined.csv", index=False, header=False)

print("done")
```
