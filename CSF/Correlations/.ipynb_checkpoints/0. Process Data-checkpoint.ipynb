```python
import os
import pandas as pd

dss = pd.read_csv("DSS.csv")
dss_map = pd.read_csv("DSS_group_map.csv", header=None)
lps = pd.read_csv("LPS.csv")
lps_map = pd.read_csv("LPS_group_map.csv", header=None)
vecpac = pd.read_csv("VECPAC.csv")
vecpac_map = pd.read_csv("VECPAC_group_map.csv", header=None)


def drop_treatment_columns(df, group_map):
    # Drop treatment columns
    # column_name:group_label dict
    mapping = {str(k): str(v) for k, v in zip(group_map.iloc[:, 0], group_map.iloc[:, 1])}

    cols_to_drop = [
        col for col in df.columns
        if mapping.get(str(col), "").strip().lower() == "treatment"
    ]
    print(cols_to_drop)
    return df.drop(columns=cols_to_drop)


def split_regions(df):
    first_row = df.iloc[[0]]
    CPX = df[df.iloc[:, 0].astype(str).str.endswith("-P")]
    CTX = df[df.iloc[:, 0].astype(str).str.endswith("-C")]
    STR = df[df.iloc[:, 0].astype(str).str.endswith("-S")]
    remaining_brain = df[df.iloc[:, 0].astype(str).str.endswith("-B")]
    return CPX, CTX, STR, remaining_brain


for folder in ["CPX", "CTX", "STR", "remaining brain"]:
    os.makedirs(folder, exist_ok=True)

# DSS
print("DSS")
dss = drop_treatment_columns(dss, dss_map)
dss_CPX, dss_CTX, dss_STR, dss_remaining_brain = split_regions(dss)
dss_CPX.to_csv("CPX/DSS_CPX.csv", index=False)
dss_CTX.to_csv("CTX/DSS_CTX.csv", index=False)
dss_STR.to_csv("STR/DSS_STR.csv", index=False)
dss_remaining_brain.to_csv("remaining brain/DSS_remaining_brain.csv", index=False)

# LPS
print("LPS")
lps = drop_treatment_columns(lps, lps_map)
lps_CPX, lps_CTX, lps_STR, lps_remaining_brain = split_regions(lps)
lps_CPX.to_csv("CPX/LPS_CPX.csv", index=False)
lps_CTX.to_csv("CTX/LPS_CTX.csv", index=False)
lps_STR.to_csv("STR/LPS_STR.csv", index=False)
lps_remaining_brain.to_csv("remaining brain/LPS_remaining_brain.csv", index=False)

# VECPAC
print("VECPAC")
vecpac = drop_treatment_columns(vecpac, vecpac_map)
vecpac_CPX, vecpac_CTX, vecpac_STR, vecpac_remaining_brain = split_regions(vecpac)
vecpac_CPX.to_csv("CPX/VECPAC_CPX.csv", index=False)
vecpac_CTX.to_csv("CTX/VECPAC_CTX.csv", index=False)
vecpac_STR.to_csv("STR/VECPAC_STR.csv", index=False)
vecpac_remaining_brain.to_csv("remaining brain/VECPAC_remaining_brain.csv", index=False)
```

```python
import pandas as pd

def pooled(tissue):
    dss = pd.read_csv(f"{tissue}/DSS_{tissue}.csv")
    lps = pd.read_csv(f"{tissue}/LPS_{tissue}.csv").iloc[:, 1:]
    vecpac = pd.read_csv(f"{tissue}/VECPAC_{tissue}.csv").iloc[:, 1:]
    
    pooled = pd.concat([dss, lps, vecpac], axis=1)
    pooled.to_csv(f"{tissue}/pooled_{tissue}.csv",index=False)

pooled("CPX")
pooled("CTX")
pooled("remaining_brain")
pooled("STR")
```
