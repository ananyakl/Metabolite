```python
import pandas as pd
import re
import os

def combine(metabolomics, transcriptomics, output_file):

    # Define folder and file path
    folder_path = "4. Combine Data Tables"
    os.makedirs(folder_path, exist_ok=True)
    output_file_path = os.path.join(folder_path, output_file)

    # Read metabolomics with no header
    meta = pd.read_csv(metabolomics, header=None)
    tran = pd.read_csv(transcriptomics, header=None)


    # The first column is the same
    meta.columns = meta.iloc[0]
    meta = meta[1:].reset_index(drop=True)
    
    tran.columns = tran.iloc[0]
    tran = tran[1:].reset_index(drop=True)
    
    combined = pd.concat([meta, tran], axis=0, join="outer", ignore_index=True)

    combined.to_csv(output_file_path, index=False, header=True)
    print("done")


# Example call
combine("2. Filter by Cortex/pooled_CTX_filter.csv", "3. Rename Metabolomics/pooled.csv", "pooled_combined.csv")
combine("2. Filter by Cortex/DSS_CTX_filter.csv", "3. Rename Metabolomics/dss.csv", "dss_combined.csv")
combine("2. Filter by Cortex/LPS_CTX_filter.csv", "3. Rename Metabolomics/lps.csv", "lps_combined.csv")
combine("2. Filter by Cortex/VECPAC_CTX_filter.csv", "3. Rename Metabolomics/vecpac.csv", "vecpac_combined.csv")
```
