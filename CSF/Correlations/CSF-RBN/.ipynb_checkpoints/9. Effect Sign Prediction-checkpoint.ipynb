```python
import pandas as pd
import numpy as np

# Load data
csf_nodes = pd.read_csv("../CSF-CPX/9. Effect Sign Prediction/CSF Effect Sign Predictions.csv") # columns: node, direction
ctx_nodes = pd.read_csv("./9. PUC/network_nodes.csv") # columns: ID, Mean Log2 Fold Change Direction (DSS)
edges = pd.read_csv("7. FDR Output/final_edges_with_fdr.csv") # columns: Gene, Metabolite, Sign

# Build lookup maps
# Gene -> direction, from network_nodes
gene_direction_map = ctx_nodes.set_index("ID")["Mean Log2 Fold Change Direction (DSS)"].to_dict()

# Metabolite -> direction, from csf_nodes
metabolite_direction_map = csf_nodes.set_index("node")["direction"].to_dict()

# Map values onto edges
edges["ctx_direction"] = edges["Gene"].map(gene_direction_map)
edges["cpx_direction"] = edges["Metabolite"].map(metabolite_direction_map)

# Reorder so the new columns sit right next to their source columns
cols = list(edges.columns)
cols.remove("ctx_direction")
cols.remove("cpx_direction")

gene_idx = cols.index("Gene")
cols.insert(gene_idx + 1, "ctx_direction")

# recompute Metabolite's index since the list shifted after inserting ctx_direction
metabolite_idx = cols.index("Metabolite")
cols.insert(metabolite_idx + 1, "cpx_direction")

edges = edges[cols]

# Sanity check: report any genes/metabolites that didn't find a match
#missing_genes = edges.loc[edges["ctx_direction"].isna(), "Gene"].unique()
#missing_metabolites = edges.loc[edges["cpx_direction"].isna(), "Metabolite"].unique()
#if len(missing_genes) > 0:
#    print(f"Warning: {len(missing_genes)} gene(s) not found in network_nodes: {missing_genes}")
#if len(missing_metabolites) > 0:
#    print(f"Warning: {len(missing_metabolites)} metabolite(s) not found in csf_nodes: {missing_metabolites}")

# Check for FDR threshold
edges["FDR < 0.15"] = (edges["Pooled FDR"] < 0.15).where(edges["Pooled FDR"].notna())

# Count non-NaN treatment groups
edges["non-NaN_treatment_groups"] = edges[["VECPAC r", "DSS r", "LPS r"]].notna().sum(axis=1).astype("Int64")



# ---------------------------------- COMPUTE PREDICTED EFFECT SIGN ----------------------------------
edges["predicted_effect_sign"] = edges["ctx_direction"] * edges["cpx_direction"]


# ---------------------------------- COMPUTE POOLED SIGN --------------------------------------------
edges["pooled_sign"] = edges["Sign"].map({"+": 1.0, "-": -1.0})

# ---------------------------------- COMPUTE WITHIN MODEL SIGN --------------------------------------
signs = np.sign(edges[["VECPAC r", "DSS r", "LPS r"]])

edges["within_model_sign"] = signs.max(axis=1).where(
    (signs.min(axis=1) == signs.max(axis=1))
    & signs.notna().any(axis=1)
)


# ---------------------------------- Comparisons --------------------------------------
edges["within_model = pooled"] = np.sign(edges["within_model_sign"]).where(
    (np.sign(edges["within_model_sign"]) == np.sign(edges["pooled_sign"]))
    & edges["within_model_sign"].notna()
    & edges["pooled_sign"].notna()
)
edges["number within_model = pooled_n"] = (edges[["within_model_sign", "pooled_sign"]].notna().sum(axis=1))


edges["within_model = predicted"] = np.sign(edges["within_model_sign"]).where(
    (np.sign(edges["within_model_sign"]) == np.sign(edges["predicted_effect_sign"]))
    & edges["within_model_sign"].notna()
    & edges["predicted_effect_sign"].notna()
)
edges["number within_model = predicted"] = (edges[["within_model_sign", "predicted_effect_sign"]].notna().sum(axis=1))


edges["predicted = pooled"] = np.sign(edges["predicted_effect_sign"]).where(
    (np.sign(edges["predicted_effect_sign"]) == np.sign(edges["pooled_sign"]))
    & edges["predicted_effect_sign"].notna()
    & edges["pooled_sign"].notna()
)
edges["number predicted = pooled"] = (edges[["predicted_effect_sign", "pooled_sign"]].notna().sum(axis=1))


signs = np.sign(edges[["within_model_sign", "predicted_effect_sign", "pooled_sign"]])
edges["all_signs_agree"] = signs.max(axis=1).where(
    (signs.min(axis=1) == signs.max(axis=1))
    & signs.notna().any(axis=1)
)
edges["number all_signs_agree"] = signs.notna().sum(axis=1).where(edges["all_signs_agree"].notna())


# ---------------------------------- Rename --------------------------------------
edges["Type_a"] = "CSF"
edges["Type_b"] = "RBN"

edges = edges.rename(columns={
    "Metabolite": "CSF_name",
    "cpx_direction": "direction_CSF",
    "Type_a":"type_CSF",
    "Gene": "brain_name",
    "ctx_direction": "direction_brain",
    "Type_b":"type_brain",
    "Consistent":"Pooled_r_valid"
})

# Select columns -----------------------------------
edges = edges[[
    "brain_name",
    "direction_brain",
    "type_brain",
    "CSF_name",
    "direction_CSF",
    "type_CSF",
    "VECPAC r",
    "VECPAC n",
    "LPS r",
    "LPS n",
    "DSS r",
    "DSS n",
    "non-NaN_treatment_groups",
    "Pooled r",
    "Pooled n",
    "Pooled_r_valid",
    "Pooled FDR",
    "FDR < 0.15",
    "within_model_sign",
    "pooled_sign",
    "predicted_effect_sign",
    "within_model = pooled",
    "number within_model = pooled_n",
    "within_model = predicted",
    "number within_model = predicted",
    "predicted = pooled",
    "number predicted = pooled",
    "all_signs_agree",
    "number all_signs_agree"
]]

# Save -----------------------------------
output_path = "9. PUC/final_edges_with_fdr_filtered_with_directions.csv"
edges.to_csv(output_path, index=False)
print(f"Saved: {output_path}")
#print(edges.head())
```
