```python
import os
import pandas as pd

# Load data
fdr = pd.read_csv("./8. Filter FDR/full_concord_fdr<0.15_final_edges_with_fdr_filtered.csv")
new_pairs = pd.read_csv("./9. Effect Sign Prediction/new_csf-cpx_node_pairs.csv")

# Remove the trailing "-P" from the Gene column
fdr["Gene"] = fdr["Gene"].str[:-2]

# Keep only matching (Gene, Metabolite) pairs
filtered_fdr = fdr.merge(
    new_pairs,
    left_on=["Gene", "Metabolite"],
    right_on=["node1", "node2"],
    how="inner"
).drop(columns=["node1", "node2"])

# Create output directory if it doesn't exist
os.makedirs("./10. After ESP", exist_ok=True)

# Save the filtered DataFrame
filtered_fdr.to_csv("./10. After ESP/after_esp.csv", index=False)

print("done")
```
