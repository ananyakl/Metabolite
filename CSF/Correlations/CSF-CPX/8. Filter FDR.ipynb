```python
import pandas as pd
import os

# Input/output folders
input_dir = "./7. FDR Output"
output_dir = "./8. Filter FDR"

os.makedirs(output_dir, exist_ok=True)

# Files to process
datasets = [
    "full_concord",
    "partial_concord",
    "no_concord",
]

# FDR thresholds
thresholds = [0.05, 0.10, 0.15, 0.20]

for dataset in datasets:
    # Read input file
    input_file = f"{input_dir}/alt_{dataset}_final_edges_with_fdr.csv"
    df = pd.read_csv(input_file)

    for threshold in thresholds:
        # Filter rows
        df_filtered = df[df["Pooled FDR"] < threshold]

        # Save output
        output_file = (
            f"{output_dir}/alt_{dataset}_fdr<{threshold:.2f}_final_edges_with_fdr_filtered.csv"
        )
        df_filtered.to_csv(output_file, index=False)

        print(f"Saved: {output_file}")

print("Done!")
```
