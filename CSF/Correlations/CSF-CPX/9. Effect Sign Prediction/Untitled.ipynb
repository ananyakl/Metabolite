```python
import pandas as pd

# Load files
node_list = pd.read_csv("node_list.csv")  # column: node
csf_sign = pd.read_csv("CSF Effect Sign Predictions.csv")  # columns: node, direction, puc_value

# Keep all nodes from node_list and add the corresponding direction
result = node_list.merge(
    csf_sign[["node", "direction"]],
    on="node",
    how="left"
)

# Save to a new CSV
result.to_csv("node_list_with_direction.csv", index=False)

print(f"Saved {len(result)} rows to node_list_with_direction.csv")
```
