```python
import os
import pandas as pd

import itertools
import csv
from scipy import stats
import numpy as np

def metabolite(file_name):
    # Read data safely
    filtered_data = pd.read_csv(file_name, header=None, na_values=["", "NaN"], keep_default_na=True)
    
    # Get metabolite names from the first column, skipping the first 2 rows
    names = filtered_data.iloc[1:, 0].tolist()
    
    # Extract only the numeric data grid (rows 1 onwards, columns 1 onwards)
    numeric_grid = filtered_data.iloc[1:, 1:].astype(float).values
    
    # Create a fast lookup mapping: metabolite_name -> index in numeric_grid
    name_to_idx = {name: i for i, name in enumerate(names)}
        
    # Generate all 2-element combinations
    pairs = list(itertools.combinations(names, 2))
    results = []

    # Loop through pairs
    for met1, met2 in pairs:
        # Fetch pre-converted rows directly by index instead of filtering dataframe
        row1 = numeric_grid[name_to_idx[met1]]
        row2 = numeric_grid[name_to_idx[met2]]

        # Create a mask for positions where BOTH are non-NaN
        mask = ~np.isnan(row1) & ~np.isnan(row2)
        
        # Apply mask to array
        r1_masked = row1[mask]
        r2_masked = row2[mask]
        count = mask.sum()

        # If less than three paired points: let corr and pval be NaN
        if count < 3:
            corr, pval = np.nan, np.nan
        else:
            # Calculate Spearman correlation
            corr, pval = stats.spearmanr(r1_masked, r2_masked)
        
        # Append results (Fixes the indentation block bug)
        results.append((met1, met2, corr, pval, count))

    # Create output directory + output file path
    parts = os.path.basename(file_name).split("_")
    prefix = "_".join(parts[:2])
    os.makedirs("1. Metabolite Pairs", exist_ok=True)
    output_file = os.path.join("1. Metabolite Pairs", f"COMBINATIONS_{prefix}")
    
    with open(output_file, "w", newline='', encoding="utf-8") as outfile:
        writer = csv.writer(outfile)
        writer.writerow(["Metabolite 1", "Metabolite 2", "Spearman Coefficient", "p-value", "Values Counted"])
        writer.writerows(results)

    print(f"Saved {output_file} with {len(results)} metabolite pairs.")

# ------------- Run -------------
metabolite("./0. Original/pooled_data.csv")
metabolite("./0. Original/vecpac_data.csv")
metabolite("./0. Original/dss_data.csv")
metabolite("./0. Original/lps_data.csv")

```
