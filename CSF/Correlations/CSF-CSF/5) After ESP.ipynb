```python
import os
import pandas as pd

# Load data
fdr = pd.read_csv("./4. Filter FDR/final_edges_with_fdr_filtered.csv")
new_pairs = pd.read_csv("./5. After ESP/new_csf-csf_node_pairs.csv")


# Keep only matching (Gene, Metabolite) pairs
filtered_fdr = fdr.merge(
    new_pairs,
    left_on=["Gene", "Metabolite"],
    right_on=["node1", "node2"],
    how="inner"
).drop(columns=["node1", "node2"])

# Save the filtered DataFrame
filtered_fdr.to_csv("./5. After ESP/after_esp.csv", index=False)

print("done")
```

```python
comparison = fdr.merge(
    new_pairs,
    left_on=["Gene", "Metabolite"],
    right_on=["node1", "node2"],
    how="outer",
    indicator=True
)

print(comparison["_merge"].value_counts())
```
