```python
import pandas as pd
import numpy as np
from statsmodels.stats.multitest import fdrcorrection
import os

# Read files
df = pd.read_csv("./2. Filter Edges/filtered_edges.csv")
vecpac = pd.read_csv("./1. Metabolite Pairs/COMBINATIONS_vecpac_data.csv")
dss    = pd.read_csv("./1. Metabolite Pairs/COMBINATIONS_dss_data.csv")
lps    = pd.read_csv("./1. Metabolite Pairs/COMBINATIONS_lps_data.csv")
pool   = pd.read_csv("./1. Metabolite Pairs/COMBINATIONS_csf_pooled.csv")


# Get valid p-values
valid_rows = (df['Consistent'] == True)
pvals = pd.to_numeric(df.loc[valid_rows, 'Pooled P-value'], errors='coerce')
# Run FDR correction directly on the p-value array
rejected, pvals_corrected = fdrcorrection(
    pvals.values,
    alpha=0.05,
    method='indep',
    is_sorted=False
)
# Add FDR results back to the dataframe
df['Pooled FDR'] = np.nan
df.loc[valid_rows, 'Pooled FDR'] = pvals_corrected



# Build final table
final_df = pd.DataFrame({
    "Gene": df.iloc[:, 0],
    "Metabolite": df.iloc[:, 1],
    "VECPAC r": vecpac.iloc[:, 2],
    "VECPAC p": vecpac.iloc[:, 3],
    "VECPAC n": vecpac.iloc[:, 4],
    "DSS r": dss.iloc[:, 2],
    "DSS p": dss.iloc[:, 3],
    "DSS n": dss.iloc[:, 4],
    "LPS r": lps.iloc[:, 2],
    "LPS p": lps.iloc[:, 3],
    "LPS n": lps.iloc[:, 4],
    "Pooled r": pool.iloc[:, 2],
    "Pooled p": pool.iloc[:, 3],
    "Pooled n": pool.iloc[:, 4],
    "Consistent": df.iloc[:, 7],
    "Sign": df.iloc[:, 8],
    "Pooled FDR": df["Pooled FDR"],
})


os.makedirs("3. FDR Output", exist_ok=True)
final_df.to_csv("3. FDR Output/final_edges_with_fdr.csv", index=False)

print("done")
```
