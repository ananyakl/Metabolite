```python
import pandas as pd
import numpy as np
from statsmodels.stats.multitest import fdrcorrection
import os

# Read file
df = pd.read_csv("./7. FDR Output/final_edges_with_fdr.csv")

# Keep only rows where Pooled FDR < 0.05
df_filtered = df[df["Pooled FDR"] < 0.15]

# Create output folder if it doesn't exist
output_dir = "./8. Filter FDR"
os.makedirs(output_dir, exist_ok=True)

# Save filtered file
output_path = os.path.join(output_dir, "final_edges_with_fdr_filtered.csv")
df_filtered.to_csv(output_path, index=False)

print(f"Filtered file saved to: {output_path}")
print(f"Rows retained: {len(df_filtered)}")
```
