```python
import pandas as pd
import os

def filter_by_CTX(input_file, output_file=None):

    # Define folder and file path
    folder_path = "2. Filter by Striatum"
    # Create the folder if it doesn't exist
    os.makedirs(folder_path, exist_ok=True)
    output_file_path = os.path.join(folder_path, output_file)

    # Read the file
    df = pd.read_csv(input_file)

    

    # Keep first row
    first_row = df.iloc[[0]]
    # Filter rows where column "ID" ends with '-S'
    filtered_df = df[df["ID"].astype(str).str.endswith("-S")]

    print(len(filtered_df))

    node = pd.read_csv("./2. Filter by Striatum/matt_table_nodes.csv")
    node = node[node["Tissue"] == "Striatum"]     # Keep only Cortex rows
    external_ids = node["External ID"].astype(str) + "-S"    # Add "-C" to the end of every row in "External ID"
    filtered_df = filtered_df[filtered_df["ID"].isin(external_ids)]     # Keep only rows whose ID is found in the Cortex External IDs
    print(len(filtered_df))
    
    filtered_df.to_csv(output_file_path, index=False)

filter_by_CTX("./1. Filter by CTRL/VECPAC_mapped.csv", "VECPAC_CTX_filter.csv")
filter_by_CTX("./1. Filter by CTRL/DSS_mapped.csv", "DSS_CTX_filter.csv")
filter_by_CTX("./1. Filter by CTRL/LPS_mapped.csv", "LPS_CTX_filter.csv")

print("done")
```

```python
import pandas as pd

# Read the CSV files
dss = pd.read_csv("./2. Filter by Striatum/DSS_CTX_filter.csv")
lps = pd.read_csv("./2. Filter by Striatum/LPS_CTX_filter.csv")
vecpac = pd.read_csv("./2. Filter by Striatum/VECPAC_CTX_filter.csv")

# Keep all columns from one.csv,
# and drop the first column from two.csv and three.csv
result = pd.concat([dss,lps.iloc[:, 1:],vecpac.iloc[:, 1:]],axis=1)

# Save the result
result.to_csv("./2. Filter by Striatum/pooled_CTX_filter.csv", index=False)

print("Saved to concatenated.csv")
```
