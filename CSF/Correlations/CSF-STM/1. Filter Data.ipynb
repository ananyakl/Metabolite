```python
import pandas as pd
import os


def ctrl_filter(orig, filter_by):

    # Read both CSV files without headers
    orig_data = pd.read_csv(orig, header=None)
    group_map = pd.read_csv(filter_by, header=None)

    # Ensure column 0 in map is numeric (for matching)
    group_map.iloc[:, 0] = pd.to_numeric(group_map.iloc[:, 0], errors='coerce')

    # Loop over each column in original file (first row values)
    for i, val in enumerate(orig_data.iloc[0]):

        # Convert to float; skip non-numeric
        try:
            val_float = int(val)
        except ValueError:
            continue

        # Find matching row in group_map
        match = group_map[group_map.iloc[:, 0] == val_float]

        # if control
        if not match.empty:
            group_value = str(match.iloc[0, 1]).strip().lower()

            if group_value != "control":
                # DELETE this column
                orig_data.drop(columns=[i], inplace=True)
            else:
                print("Column kept")
        else:
            print("error")

    # Save filtered output
    output_path = orig.replace(".csv", "_filtered.csv")
    orig_data.to_csv(output_path, index=False, header=False)
    print(f"Saved: {output_path}")


# Run for your files
ctrl_filter("./0. Original/DSS.csv", "./0. Original/DSS_group_map.csv")
ctrl_filter("./0. Original/LPS.csv", "./0. Original/LPS_group_map.csv")
```

```python
import pandas as pd
import os

# Define folder and file path
folder_path = "1. Filter by CTRL"
# Create the folder if it doesn't exist
os.makedirs(folder_path, exist_ok=True)


def rename(input_file, output_file):
    # Read both CSV files without headers
    mouse_map = pd.read_csv('./0. Original/mouse_map.csv', header=None)
    vecpac = pd.read_csv(input_file, header=None)
    output_file_path = os.path.join(folder_path, output_file)

    for i, val in enumerate(vecpac.iloc[0]):
        # Look for a row in mouse_map where the first column matches the VECPAC value
        # Convert to float
        try:
            val_float = float(val)
        except ValueError:
            continue  # skip non-numeric values
            
        mouse_map.iloc[:, 0] = pd.to_numeric(mouse_map.iloc[:, 0], errors='coerce')
        match = mouse_map[mouse_map.iloc[:, 0] == val_float]
    #---------------------------------------------------------------------------------        
        # If a match is found, replace the VECPAC value with the second column of that row
        if not match.empty:
            #print("replaced")
            vecpac.iat[0, i] = match.iloc[0, 1]  # <-- uses the SECOND column
        else:
            print("not replaced")

    # Save the modified VECPAC file
    vecpac.to_csv(output_file_path, index=False, header=False)
    print("File saved")

rename("./0. Original/VECPAC_filtered.csv", "VECPAC_mapped.csv")
rename("./0. Original/DSS_filtered.csv", "DSS_mapped.csv")
rename("./0. Original/LPS_filtered.csv", "LPS_mapped.csv")
```
