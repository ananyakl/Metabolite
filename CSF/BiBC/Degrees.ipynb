```python
import pandas as pd
from collections import defaultdict

# Load node types (no header)
nodes = pd.read_csv(
    "cpx_brain_node_types.csv",
    header=None,
    names=["node", "type"]
)

# Load edge list
edges = pd.read_csv("cpx_brain_final_edges_table.csv")

# Dictionary: node -> type
node_type = dict(zip(nodes["node"], nodes["type"]))

# Initialize counts
degree = defaultdict(int)
stm_degree = defaultdict(int)
rbn_degree = defaultdict(int)
cpx_degree = defaultdict(int)
ctx_degree = defaultdict(int)
csf_degree = defaultdict(int)

# Iterate through edges
for _, row in edges.iterrows():
    n1 = row["node1"]
    n2 = row["node2"]

    t1 = node_type.get(n1)
    t2 = node_type.get(n2)

    # If node1 is CSF
    if t1 == "CSF":
        degree[n1] += 1

        if t2 == "STM":
            stm_degree[n1] += 1
        elif t2 == "RBN":
            rbn_degree[n1] += 1
        elif t2 == "CPX":
            cpx_degree[n1] += 1
        elif t2 == "CTX":
            ctx_degree[n1] += 1
        elif t2 == "CSF":
            csf_degree[n1] += 1

    # If node2 is CSF
    if t2 == "CSF":
        degree[n2] += 1

        if t1 == "STM":
            stm_degree[n2] += 1
        elif t1 == "RBN":
            rbn_degree[n2] += 1
        elif t1 == "CPX":
            cpx_degree[n2] += 1
        elif t1 == "CTX":
            ctx_degree[n2] += 1
        elif t1 == "CSF":
            csf_degree[n2] += 1

# Create results table
results = pd.DataFrame({
    "CSF_node": [n for n in nodes[nodes["type"] == "CSF"]["node"]],
})

results["Total_degree"] = results["CSF_node"].map(degree).fillna(0).astype(int)
results["STM_degree"] = results["CSF_node"].map(stm_degree).fillna(0).astype(int)
results["RBN_degree"] = results["CSF_node"].map(rbn_degree).fillna(0).astype(int)
results["CPX_degree"] = results["CSF_node"].map(cpx_degree).fillna(0).astype(int)
results["CTX_degree"] = results["CSF_node"].map(ctx_degree).fillna(0).astype(int)
results["CSF_degree"] = results["CSF_node"].map(csf_degree).fillna(0).astype(int)

print(results)

# Save
results.to_csv("CSF_node_degrees.csv", index=False)
```
