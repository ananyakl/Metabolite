```python
import numpy as np
import pandas as pd
import os


dss = pd.read_csv("./5. Metabolite Pairs/dss_met_pair.csv", header=None, na_values=["", "NaN"], keep_default_na=True)
lps = pd.read_csv("./5. Metabolite Pairs/lps_met_pair.csv", header=None, na_values=["", "NaN"], keep_default_na=True)
vecpac = pd.read_csv("./5. Metabolite Pairs/vecpac_met_pair.csv", header=None, na_values=["", "NaN"], keep_default_na=True)
comb = pd.read_csv("./5. Metabolite Pairs/all_met_pair.csv", header=None, na_values=["", "NaN"], keep_default_na=True)

# Remove first row from each (header)
dss = dss.iloc[1:].reset_index(drop=True)
lps = lps.iloc[1:].reset_index(drop=True)
vecpac = vecpac.iloc[1:].reset_index(drop=True)
comb = comb.iloc[1:].reset_index(drop=True)



# Get first two columns with transcriptomics and metabolomics names
first_two = dss.iloc[:, :2]

# Get third column (Spearmann Coefficients) for each file
vecpac_coef = vecpac.iloc[:, 2].reset_index(drop=True)
lps_coef    = lps.iloc[:, 2].reset_index(drop=True)
dss_coef    = dss.iloc[:, 2].reset_index(drop=True)
comb_coef   = comb.iloc[:, 2].reset_index(drop=True)

same_sign = []
sign = []
for i in range(len(vecpac_coef)):
    
    v = vecpac_coef.iloc[i]
    l = lps_coef.iloc[i]
    d = dss_coef.iloc[i]
    p = comb_coef.iloc[i]

    # try converting it to float and check if NaN
    try:
        vals = [float(v), float(l), float(d), float(p)]
    except:
        same_sign.append(False)
        continue 
    if any(np.isnan(x) for x in vals):
        same_sign.append(False)
        continue

    
    if (float(v) > 0 and float(l) > 0 and float(d) > 0 and float(p) > 0):
        same_sign.append(True)
    elif (float(v) < 0 and float(l) < 0 and float(d) < 0 and float(p) < 0):
        same_sign.append(True)
    else:
        same_sign.append(False)


new_df = pd.DataFrame({
    "Gene": first_two.iloc[:, 0],
    "Metabolite": first_two.iloc[:, 1],
    "VECPAC r": vecpac_coef,
    "LPS r": lps_coef,
    "DSS r": dss_coef,
    "Pooled Data r": comb_coef,
    "Pooled P-value": vecpac.iloc[:, 3],
    "Consistent": same_sign,
})


os.makedirs("6. Filter Edges", exist_ok=True)  
output_file_path = os.path.join("6. Filter Edges", "filtered_edges.csv") 
new_df.to_csv(output_file_path, index=False)
print("done")


```
