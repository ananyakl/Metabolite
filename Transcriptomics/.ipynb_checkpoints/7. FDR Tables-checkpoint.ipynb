```python
import pandas as pd
import numpy as np
from statsmodels.stats.multitest import fdrcorrection

# ------------------------
# Load all CSV files
# ------------------------
df = pd.read_csv("./6. Filter Edges/filtered_edges.csv")
vecpac = pd.read_csv("./5. Metabolite Pairs/vecpac_met_pair.csv")
dss    = pd.read_csv("./5. Metabolite Pairs/dss_met_pair.csv")
lps    = pd.read_csv("./5. Metabolite Pairs/lps_met_pair.csv")
pool   = pd.read_csv("./5. Metabolite Pairs/all_met_pair.csv")


# Get FDR
is_valid = ~df['Pooled P-value'].isna()
rejected, pvals_corrected = fdrcorrection(df.loc[is_valid, 'Pooled P-value'], alpha=0.05, method='indep', is_sorted=False)
df["Pooled FDR"] = np.nan
df.loc[is_valid, "Pooled FDR"] = pvals_corrected

# Build final table
final_df = pd.DataFrame({
    "Metabolite 1": df.iloc[:, 0],
    "Metabolite 2": df.iloc[:, 1],
    "VECPAC r": vecpac.iloc[:, 2],
    "VECPAC p": vecpac.iloc[:, 3],
    "VECPAC n": vecpac.iloc[:, 4],
    "DSS r": dss.iloc[:, 2],
    "DSS p": dss.iloc[:, 3],
    "DSS n": dss.iloc[:, 4],
    "LPS r": lps.iloc[:, 2],
    "LPS p": lps.iloc[:, 3],
    "LPS n": lps.iloc[:, 4],
    "Pooled r": pool.iloc[:, 2],
    "Pooled p": pool.iloc[:, 3],
    "Pooled n": pool.iloc[:, 4],
    "Consistent?": df.iloc[:, 7],
    "Sign?": df.iloc[:, 8],
    "Pooled FDR": df["Pooled FDR"]
})


os.makedirs("7. FDR Output", exist_ok=True)
final_df.to_csv("7. FDR Output/final_edges_with_fdr.csv", index=False)

print("done")
```
