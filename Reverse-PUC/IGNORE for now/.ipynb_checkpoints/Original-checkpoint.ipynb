```python
import networkx as nx
import matplotlib.pyplot as plt

G_a = nx.barabasi_albert_graph(7, m=2, seed=10)
G_b = nx.barabasi_albert_graph(4, m=3, seed=10)

# Relabel nodes so they don't collide
G_a = nx.relabel_nodes(G_a, lambda x: f"a{x}")
G_b = nx.relabel_nodes(G_b, lambda x: f"b{x}")

G = nx.compose(G_a, G_b)

cross_edges = [
    ("a0", "b0"),
    ("a1", "b1"),
    ("a0", "b1"),
    ("a0", "b2")
]
G.add_edges_from(cross_edges)
```

```python
pos_a = nx.spring_layout(G_a, seed=1)
pos_b = nx.spring_layout(G_b, seed=1)

# shift G_b to the right
pos_b = {k: (v[0] + 4, v[1]) for k, v in pos_b.items()}

pos = {**pos_b, **pos_a}

pos = nx.circular_layout(G)
nx.draw(G, pos=pos, with_labels=True)

plt.show()
```

```python
sources = [n for n in G.nodes if n.startswith("b")]

distances = nx.multi_source_shortest_path_length(G, sources)

from collections import defaultdict

layers = defaultdict(list)

for node, d in distances.items():
    layers[d].append(node)

for k in sorted(layers):
    print(f"L_{k}: {layers[k]}")

pos = {}

for layer, nodes in layers.items():
    for i, node in enumerate(nodes):
        pos[node] = (i, -layer)  # negative so L0 is on top

nx.draw(
    G,
    pos,
    with_labels=True,
    node_color=[
        "lightgreen" if n.startswith("b") else "lightblue" for n in G.nodes
    ]
)
plt.show()
```
