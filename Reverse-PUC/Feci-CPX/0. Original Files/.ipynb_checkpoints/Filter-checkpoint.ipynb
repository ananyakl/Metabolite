CPX_node

```python
import pandas as pd
from pathlib import Path

# Read the CSV with header
df = pd.read_csv("cpx_node.csv")

# Identify first column
first_col = df.columns[0]

# Keep only rows where the first column ends with "-P"
filtered_df = df[df[first_col].astype(str).str.endswith("-P")].copy()

# Remove trailing "-P" from the first column values
filtered_df[first_col] = filtered_df[first_col].astype(str).str.replace(r"-P$", "", regex=True)

# Ensure output directory exists
output_dir = Path("../1. Filtered Files")
output_dir.mkdir(parents=True, exist_ok=True)

# Save the filtered table
output_path = output_dir / "cpx_nodes.csv"
filtered_df.to_csv(output_path, index=False)

print(f"Filtered file saved to: {output_path}")
```

Feci-Feci -edges

```python
import pandas as pd
from pathlib import Path
import numpy as np

# Read the CSV with header
df = pd.read_csv("feci-feci.csv")


# Filter rows where column 13 is TRUE
df = df[df.iloc[:, 12] == True]

# after filtering, add a column to the right called 'Sign'
# Check if 'DSS r' is positive or negative. If null, if 'LPS r' is positive or negative. 'Sign' is -1 if negative, and +1 if postiive
df['Sign'] = np.where(
    df['DSS r'].notna(), 
    np.where(df['DSS r'] > 0, 1, -1),  # Use DSS r if available
    np.where(df['LPS r'] > 0, 1, -1)   # Otherwise use LPS r
)

df = df[(df["FDR"] <= 0.05) & (df[["VECPAC p-values", "DSS p-values", "LPS p-values"]].max(axis=1) <= 0.2)]





# Keep only 1st, 2nd, and 16th columns
# (indices 0, 1, and 15)
final_df = df.iloc[:, [0, 1, 15]].copy()

# Ensure output directory exists
output_dir = Path("../1. Filtered Files")
output_dir.mkdir(parents=True, exist_ok=True)

# Save the final CSV
output_path = output_dir / "feci-feci.csv"
final_df.to_csv(output_path, index=False)

print(f"Filtered file saved to: {output_path}")
```

```python
import pandas as pd
from pathlib import Path

# Read the CSV with header
df = pd.read_csv("feci-cpx.csv")

#Filterr those that are not null
df = df[df['Pooled FDR'].notna()]
df = df[df["Pooled FDR"] <= 0.1]

df['Sign'] = df['Sign'].map({'+': 1, '-': -1})


# Keep only 1st, 2nd, and 16th columns (0-based: 0, 1, 15)
final_df = df.iloc[:, [0, 1, 15]].copy()


# Remove ONLY trailing "-P" from all cells in the first column
first_col = final_df.columns[0]
final_df[first_col] = (
    final_df[first_col]
    .astype(str)
    .str.replace(r"-P$", "", regex=True)
)

# Ensure output directory exists
output_dir = Path("../1. Filtered Files")
output_dir.mkdir(parents=True, exist_ok=True)

# Save the final CSV
output_path = output_dir / "feci-cpx.csv"
final_df.to_csv(output_path, index=False)

print(f"Filtered file saved to: {output_path}")
```
