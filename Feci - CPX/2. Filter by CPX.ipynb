```python
import pandas as pd
import os

def filter_by_CPX(input_file, output_file=None):
    
    # Define folder and file path
    folder_path = "2. Filter by Choroid Plexus"
    # Create the folder if it doesn't exist
    os.makedirs(folder_path, exist_ok=True)
    output_file_path = os.path.join(folder_path, output_file)
    
    # Read the file (no header)
    df = pd.read_csv(input_file, header=None)
    df.iloc[0] = df.iloc[0].apply(
        lambda x: str(int(x)) if isinstance(x, (int, float)) else x
    )

    print(df)

    # Keep first row
    first_row = df.iloc[[0]]
    # Filter rows where last character of column 0 ends with 'P'
    filtered_df = df[df[0].astype(str).str.endswith('P')]
    filtered_df = pd.concat([first_row, filtered_df], ignore_index=True)

    filtered_df.to_csv(output_file_path, index=False, header=False)

filter_by_CPX("./1. Filter by CTRL/VECPAC_mapped.csv","VECPAC_CPX_filter.csv")
filter_by_CPX("./1. Filter by CTRL/DSS_mapped.csv","DSS_CPX_filter.csv")
filter_by_CPX("./1. Filter by CTRL/LPS_mapped.csv","LPS_CPX_filter.csv")

print("done")
```
